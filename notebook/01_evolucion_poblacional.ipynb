## Actividad 3 — Estadística descriptiva

**Propósito:** Aplicar medidas estadísticas básicas a una variable poblacional.

Se selecciona el año **2023** (último año con datos históricos, ya que el rango histórico abarca de 1951 a 2023) y se filtran los registros correspondientes sobre la variable `Annual change in population`.

```python
col = "Annual change in population"

# Se filtran solo los datos históricos (no nulos) para el año seleccionado
df_2023 = df[(df["Year"] == 2023) & (df[col].notnull())]
df_2023.head()
```

```python
media = df_2023[col].mean()
mediana = df_2023[col].median()
varianza = df_2023[col].var()
desviacion = df_2023[col].std()
minimo = df_2023[col].min()
maximo = df_2023[col].max()

print("Media:", media)
print("Mediana:", mediana)
print("Varianza:", varianza)
print("Desviación estándar:", desviacion)
print("Mínimo:", minimo)
print("Máximo:", maximo)
```

```python
entidad_min = df_2023.loc[df_2023[col].idxmin(), "Entity"]
entidad_max = df_2023.loc[df_2023[col].idxmax(), "Entity"]

print("Entidad con el valor mínimo:", entidad_min)
print("Entidad con el valor máximo:", entidad_max)
```
